```python
class Employee:
    def __init__(self,employee_id,employee_name):
        self.employee_id=employee_id
        self.employee_name=employee_name

    def display(self):
        print(self.employee_id)
        print(self.employee_name)


emp1=Employee(1,"p")
emp1.display()
emp2=Employee(2,"b")
emp2.display()
emp3=Employee(3,"c")
emp3.display()
```
